```python
op_object_dict = {}
```

```python
import pandas as pd

for opid in list_opids:
    relation_name = df_explain_stream[(df_explain_stream['EXPLAIN_TIME'] == explain_time) 
                                      & (df_explain_stream['TARGET_ID'] == opid)]['OBJECT_NAME'].values[0]
    # Convert relation_name to a pandas Series and check if it's not null
    if not pd.isnull(relation_name):
        op_object_dict[opid] = relation_name
        # print('opid: {}, OBJECT: {}'.format(opid, relation_name))
```

```python
op_object_dict
```

```python
if op_object_dict.get(1) is not None:
    Plan['Relation Name'] = op_object_dict.get(1)
```
