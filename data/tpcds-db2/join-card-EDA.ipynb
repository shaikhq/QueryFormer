# Phase 2 Analysis - Combining relevant data elements from multiple tables


```python
df_names_filtered = ["df_operator_filtered", "df_actuals_filtered", "df_predicate_filtered", "df_stream_filtered"]
list_of_dfs = [df_operator, df_actuals, df_predicate, df_stream]
```

```python
df_operator.shape
```

```python
df_predicate.shape
```

```python
df_stream.shape
```

```python
filtered_dfs = {}

for df, name in zip(list_of_dfs, df_names_filtered):
    filtered_df = df[df['EXPLAIN_TIME'] == query1_ts]
    filtered_dfs[name] = filtered_df
```

```python
filtered_dfs.keys()
```

```python
for name, df in filtered_dfs.items():
    print(f"The shape of {name} is {df.shape}")
```

```python
## filtering columns
list_of_column_names = [operator_cols, actuals_cols, predicate_cols, stream_cols]
filtered_dfs = {name: df[column_names] for (name, df), column_names in zip(filtered_dfs.items(), list_of_column_names)}
```

```python
for name, df in filtered_dfs.items():
    print(f"The shape of {name} is {df.shape}")
```

## filter and group predicate df

```python
temp_df = filtered_dfs['df_predicate_filtered']
temp_df = temp_df[temp_df['OPERATOR_ID'].isin(filtered_dfs['df_operator_filtered']['OPERATOR_ID'])]
```

```python
temp_df
```

```python
## flatten predicate df
# Assuming temp_df is your DataFrame
df_grouped = temp_df.groupby('OPERATOR_ID').agg({
    'PREDICATE_TEXT': list,
    'FILTER_FACTOR': list
}).reset_index()

# Create new columns for each pair of 'PREDICATE_TEXT' and 'FILTER_FACTOR'
df_grouped['PREDICATE_TEXT_1'] = df_grouped['PREDICATE_TEXT'].apply(lambda x: x[0])
df_grouped['FILTER_FACTOR_1'] = df_grouped['FILTER_FACTOR'].apply(lambda x: x[0])

# Drop the original 'PREDICATE_TEXT' and 'FILTER_FACTOR' columns
df_grouped = df_grouped.drop(columns=['PREDICATE_TEXT', 'FILTER_FACTOR'])
```

```python
df_grouped
```

```python
filtered_dfs['df_predicate_filtered'] = df_grouped
```
